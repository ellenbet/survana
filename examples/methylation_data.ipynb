```python
import pandas as pd
from sklearn.impute import KNNImputer

data_subset = pd.read_csv(
    "../data/all_norm_meth_1000_random_transposed.tsv", sep="\t"
)
data_subset = data_subset.set_index("PATIENT_ID")
data_subset.index.name = "PATIENT_ID"

nan_percentages = data_subset.isna().mean().mul(100)
print(f"Overall NaN percentage: {data_subset.isna().mean().mean() * 100:.2f}%")
print("NaN percentage per column:")
print(nan_percentages[nan_percentages > 0].sort_values(ascending=False))

n_neighbours = 5
numeric_columns = data_subset.select_dtypes(include="number").columns
imputer = KNNImputer(n_neighbors=n_neighbours)
data_subset[numeric_columns] = imputer.fit_transform(
    data_subset[numeric_columns]
)
data_subset.to_csv("../data/all_norm_meth_1000_random_knn-imputed.csv")
```
